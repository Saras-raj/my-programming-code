# LINEAR GRAPH

```python
x = [1,2,3,4]
y = [5,6,7,8]

plt.plot(x,y)
```

```python
x=[10,20,30,40,50]
y=[60,70,80,90,100]
plt.plot(x,y)
plt.show()
```

HOW TO CHANGE THE COLOR

```python
x=[10,20,30,40,50]
y=[60,70,80,90,100]
z= 'red'
plt.plot(x,y,z)
plt.show()
```

```python
x= [5,2,9,4,7]
y= [10,5,8,4,2]
plt.plot(x,y)
plt.show()
```

```python
x= [5,2,9,4,7]
y= [10,5,8,4,2]
plt.plot(x,y, marker = 'o')
plt.show()
```

```python
x= [5,2,9,4,7]
y= [10,5,8,4,2]
plt.plot(x,y)
plt.plot(y,marker = ">")
plt.show()
```

```python
x= [5,2,9,4,7]
y= [10,5,8,4,2]
plt.plot(x,y)
plt.plot(y, marker = ">")
plt.plot(x, marker = "o")
plt.show()
```

# BAR GRAPH

```python
x= [1,2,3,4]
y= [5,6,7,8]
plt.bar(x,y)
plt.show()
```

```python
x= [1,2,3,4]
y= [5,6,7,8]
z= ['r','g','b','y']
plt.bar(x,y, color = z)
plt.show()
```

# Scatter Plot

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,1,8]

plt.scatter(x,y)
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,1,8]
plt.title("scatter plot")
plt.scatter(x,y)
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,1,8]
plt.title("scatter plot")
plt.xlabel("month")
plt.ylabel("number")
plt.scatter(x,y)
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,1,8]
plt.title("scatter plot")
plt.xlabel("month")
plt.ylabel("number")
plt.scatter(x,y, color = 'r')
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,1,8]
plt.title("scatter plot")
plt.xlabel("month")
plt.ylabel("number")

z = ['red','magenta','orange','blue','brown','pink'] 
plt.scatter(x,y, color = z)
plt.show()
```

```python
x = [1,2,3,4,5,6]
y = [4,2,3,5,1,8]
plt.title("scatter plot")
plt.xlabel("month")
plt.ylabel("number")

z = ['red','magenta','orange','blue','brown','pink'] 
plt.scatter(x,y, color = z ,s=150)
plt.show()
```

# Pie Chart

```python
import matplotlib.ppylot as plt
```

```python
x = [10,20,30,40]
y = ['english','hindi','math','science']

plt.pie(x, labels=y)
plt.show()
```

```python
x = [10,20,30,40]
y = ['english','hindi','math','science']
z = ['yellow','magenta','brown','aqua']
plt.pie(x, labels=y, colors = z)
plt.show()
```

```python
x = [10,20,30,40]
y = ['english','hindi','math','science']

plt.pie(x, labels=y)
plt.legend()
plt.show()
```
